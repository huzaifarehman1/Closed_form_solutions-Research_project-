The problem we are trying to solve is that for a dataset `X` used to predict output vector `Y`, a normal Neural Network Randomly initializes all Layers and then we train the neural network to reduce the Loss defined by the `Loss function`. However, we know that a simple Linear Regression model can be solved by using the `First Derivative test`. We solve for the critical point to find the Minimum this is called `Closed form solution for Linear regression`.The purpose of this research is to take inspiration from this idea to fully or partially eliminate the training phase of a neural network by substituting it for a Closed form solution

### Closed Form solution for Linear Regression


Consider a linear regression model:

$$
\hat{\mathbf{y}} = \mathbf{X}\boldsymbol{\beta}
$$

where **X** is the design matrix, **y** is the target vector, and **β** is the vector of model parameters.

The parameters are estimated by minimizing the sum of squared errors:

$$
\mathcal{L}(\boldsymbol{\beta})
=
\|\mathbf{y}-\mathbf{X}\boldsymbol{\beta}\|_2^2
$$

Expanding the objective function gives:

$$
\mathcal{L}(\boldsymbol{\beta})
=
(\mathbf{y}-\mathbf{X}\boldsymbol{\beta})^T
(\mathbf{y}-\mathbf{X}\boldsymbol{\beta})
$$

Taking the gradient with respect to **β**:

$$
\nabla_{\boldsymbol{\beta}}\mathcal{L}
=
-2\mathbf{X}^T\mathbf{y}
+
2\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}
$$

At the minimum, the gradient is equal to zero:

$$
-2\mathbf{X}^T\mathbf{y}
+
2\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}
=0
$$

Rearranging produces the **normal equations**:

$$
\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}
=
\mathbf{X}^T\mathbf{y}
$$

If $\mathbf{X}^T\mathbf{X}$ is invertible, multiplying both sides by $(\mathbf{X}^T\mathbf{X})^{-1}$ yields the closed-form solution:

$$
\boldsymbol{\beta}^{*}
=
(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}
$$
$$
\boxed{
\boldsymbol{\beta}^{*}
=
(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}
}
$$

Therefore, the predicted values are:

$$
\boxed{
\hat{\mathbf{y}}
=
\mathbf{X}(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}
}
$$

The matrix

$$
\mathbf{H}
=
\mathbf{X}(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T
$$

is called the **hat matrix**, because it maps the observed target vector to the fitted values:

$$
\hat{\mathbf{y}}=\mathbf{H}\mathbf{y}
$$

If $\mathbf{X}^T\mathbf{X}$ is singular and therefore cannot be inverted, the Moore–Penrose pseudoinverse can be used instead:

$$
\boxed{
\boldsymbol{\beta}^{*}
=
\mathbf{X}^{+}\mathbf{y}
}
$$

This formulation provides a least-squares solution even when the columns of the design matrix are linearly dependent.

### Structure of Neural Network


A neural network is composed of multiple layers, where each layer typically consists of two fundamental operations: an affine transformation followed by a nonlinear activation function. The affine transformation is analogous to the linear component of linear regression, where the input is multiplied by a weight matrix and a bias vector is added. The activation function then introduces nonlinearity, allowing the network to model complex relationships that cannot be represented by a purely linear model.

Let (NN(X)) denote a neural network that takes (X) as its input, where each row of (X) represents a single data point. A single layer of the neural network can be defined as

$$
\mathrm{Layer}(X) = \sigma(XW + b),
$$

where (W) is the weight matrix, (b) is the bias vector, and (\sigma) is a nonlinear activation function applied element-wise.

The affine transformation (XW+b) determines a linear transformation of the input features, while the activation function (\sigma) introduces the nonlinearity necessary for the network to learn more complex functions. Without the activation function, stacking multiple such layers would still result in an overall linear transformation and therefore would not provide the expressive power associated with neural networks.

A neural network can contain multiple layers, with the output of one layer serving as the input to the next. For a network with (L) layers, the computation can be written recursively as

$$
H^{(0)} = X,
$$

$$
H^{(l)} = \sigma_l\left(H^{(l-1)}W^{(l)} + b^{(l)}\right),
\qquad l=1,\ldots,L,
$$

where (W^{(l)}) and (b^{(l)}) are the weight matrix and bias vector of layer (l), respectively, and (\sigma_l) is its activation function. The final output of the neural network is therefore

$$
NN(X) = H^{(L)}.
$$

By composing multiple affine transformations with nonlinear activation functions, neural networks can represent highly complex nonlinear mappings from inputs to outputs.
